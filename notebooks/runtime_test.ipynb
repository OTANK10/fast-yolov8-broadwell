```python
import onnx
from pathlib import Path
import sys
sys.path.append("..")
import os
import onnxruntime as ort
```

```python
BASE_PATH = Path("..")
MODEL_PATH = BASE_PATH / "data" / "models" / "yolov8n_b1.onnx"

onnx_model = onnx.load(MODEL_PATH)
onnx.checker.check_model(onnx_model)
```

```python
from ultralytics.data.augment import LetterBox
import cv2
import matplotlib.pyplot as plt

DATA_PATH = BASE_PATH / "val2017"
FILE_IDX = 0 
filenames = sorted(os.listdir(DATA_PATH))
image_data = cv2.imread(DATA_PATH / filenames[FILE_IDX])

lb = LetterBox(new_shape=(640, 640))
boxxed = lb(image=image_data)
```

```python
ort_sess = ort.InferenceSession(MODEL_PATH, providers=["CPUExecutionProvider"])
blob = cv2.dnn.blobFromImages([boxxed], scalefactor=1/255, swapRB=True)
outputs = ort_sess.run(None, {'images': blob})
```
